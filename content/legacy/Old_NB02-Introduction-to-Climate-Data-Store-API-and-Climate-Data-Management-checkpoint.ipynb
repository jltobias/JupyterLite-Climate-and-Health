## Building our first CDSAPI requests to leverage the Copernicus Climate Data Store datasets
 - How can we quickly leverage data using the Copernicus Climate Data Store API (cdsapi) and xarray ?
 - Let's see if we can download some historical temperature data trends with monthly average temperatures.
 - This will be a good first test to see if we can leverage climate data with python using the cdsapi


```python
# first we need to import the python library, cdsapi
import cdsapi

# Next we need to specify the dataset
dataset = "reanalysis-era5-single-levels-monthly-means"

# Next- we need to form an API request 
request = {
    "product_type": "monthly_averaged_reanalysis",
    "variable": "2m_temperature",
    "year": [
        "1979", "1980", "1981", "1982", "1983", "1984", "1985", "1986", "1987", "1988", "1989", "1990", "1991", "1992", "1993",
        "1994", "1995", "1996", "1997", "1998", "1999", "2000", "2001", "2002", "2003", "2004", "2005", "2006", "2007", "2008",
        "2009", "2010", "2011", "2012", "2013", "2014", "2015", "2016", "2017", "2018", "2019", "2020", "2021",
    ],
    "month": [
        "01", "02", "03", "04", "05", "06", "07", "08", "09", "10", "11", "12"
    ],
    "time": "00:00",
    "format": "netcdf"
}
# Next we initialize the client
client = cdsapi.Client()

# finally, we retrieve and download the netcdf4 file with the requested climate data
client.retrieve(dataset, request, "era5_temperature_data.nc")
```

## Now we are going to use xarray to load the NETCDF4 file (era_temperature_data.nc) and create an image file called global_annual_mean_temperature.png

```python
# first we need to import the python library for xarray and give it an alias of xr
# we do the same for matplotlib.pyplot and give it an alias of plt
import xarray as xr
import matplotlib.pyplot as plt

# Function to load the data, process it, and plot the results
def plot_global_annual_mean_temperature(nc_file):
    # Load the dataset
    data = xr.open_dataset(nc_file)
    print(data)
    
    # Convert the temperature from Kelvin to Celsius
    temperature_data = data['t2m'] - 273.15
    
    # Group by year and compute the mean temperature for each year
    mean_temperature_spatial = temperature_data.mean(dim=['latitude', 'longitude'])
    mean_temperature_by_year = mean_temperature_spatial.groupby('time.year').mean(dim='time')
    
    # Plotting the results
    plt.figure(figsize=(10, 5))
    plt.plot(mean_temperature_by_year['year'], mean_temperature_by_year.values, marker='o')
    plt.xticks(range(1980, 2021, 5))  # Adjusted to match the data range

    # Add labels and title
    plt.xlabel('Year')
    plt.ylabel('Temperature (°C)')
    plt.title('Global Annual Mean Temperature Increases (1979-2020)')
    plt.grid(True)
    plt.tight_layout()

    # Save the plot to a file
    plt.savefig("global_annual_mean_temperature.png", dpi=300)
    print("Plot saved as global_annual_mean_temperature.png")

# Example usage
nc_file = 'era5_monthly_t2m_eur.nc'
plot_global_annual_mean_temperature(nc_file)
```

## First, let's retrieve the data we need using the Climate Data Store API (cdsapi)

Step 1: Import Libraries
First, import the necessary libraries for working with NETCDF files and plotting and then use the cdsapi to retrieve the netcdf4 file

```python
# import the needed python libraries for the api request using the RCP 2.6 scenario
import cdsapi
import netCDF4 as nc
import numpy as np
import folium
import os
import zipfile

# specify the dataset
dataset = "multi-origin-c3s-atlas"
request = {
    "origin": "cmip6",
    "experiment": "ssp1_2_6",
    "domain": "global",
    "period": "2025-2100",
    "variable": "monthly_count_of_days_with_bias_adjusted_maximum_temperature_above_35_c"
    #"format": "zip"  # Explicitly requesting a ZIP format in case it comes as a compressed file
}

# initialize the client
client = cdsapi.Client()

# Define the output ZIP file name
zip_file_name = "tx35ba_CMIP6_ssp126_mon_202501-210012.zip"

# Download the file using cdsapi
client.retrieve(dataset, request, zip_file_name)
```

### Next we need to extract the netcdf4 file from the downloaded zipfile.

```python
# Check if the zipfile has been downloaded
if not os.path.exists(zip_file_name):
    print(f"Error: File '{zip_file_name}' not found.")
    exit()

# 2. Check if the file is a ZIP archive and extract it if necessary
if zipfile.is_zipfile(zip_file_name):
    with zipfile.ZipFile(zip_file_name, 'r') as zip_ref:
        zip_ref.extractall(".")  # Extract contents to the current directory
    # Find the extracted NetCDF file
    extracted_files = [f for f in os.listdir(".") if f.endswith('.nc')]
    if not extracted_files:
        print("Error: No NetCDF (.nc) file found in the extracted archive.")
        exit()
    else:
        file_name = extracted_files[0]  # Use the first extracted NetCDF file
        print(f"NetCDF file extracted: {file_name}")
else:
    print(f"Error: The file '{zip_file_name}' is not a valid ZIP archive.")
    exit()
```

### We may want to do this for any / all zipfiles in the directory 

```python
import os
import zipfile

# Function to process all ZIP files in the current directory
def process_all_zip_files():
    # List all ZIP files in the current directory
    zip_files = [f for f in os.listdir(".") if f.endswith('.zip')]
    
    if not zip_files:
        print("Error: No ZIP files found in the current directory.")
        return

    extracted_nc_files = []  # List to store the names of all extracted NetCDF files
    
    for zip_file_name in zip_files:
        print(f"Processing ZIP file: {zip_file_name}")
        # Check if the file is a ZIP archive
        if zipfile.is_zipfile(zip_file_name):
            with zipfile.ZipFile(zip_file_name, 'r') as zip_ref:
                # Extract all contents to the current directory
                zip_ref.extractall(".")
            
            # Find the extracted NetCDF files
            extracted_files = [f for f in os.listdir(".") if f.endswith('.nc')]
            if not extracted_files:
                print(f"Warning: No NetCDF (.nc) files found in the archive '{zip_file_name}'.")
            else:
                # Append the names of extracted NetCDF files to the list
                extracted_nc_files.extend(extracted_files)
                print(f"NetCDF files extracted from '{zip_file_name}': {extracted_files}")
        else:
            print(f"Error: The file '{zip_file_name}' is not a valid ZIP archive.")
    
    # Print all extracted NetCDF files
    if extracted_nc_files:
        print("\nAll extracted NetCDF files:")
        for nc_file in extracted_nc_files:
            print(nc_file)
    else:
        print("No NetCDF (.nc) files were extracted from the ZIP archives.")

# Run the function
process_all_zip_files()
```

# What if there are more than 30 consecutive days of deadly heat ?  We need  Seasonal plots (Winter, Spring, Summer, Autumn)

```python
# Import required libraries
import xarray as xr
import hvplot.xarray  # for plotting with hvplot
import geoviews as gv
import geopandas as gpd
import cartopy.crs as ccrs
import holoviews as hv

# Enable the Bokeh and Geoviews extension for interactive plots
hv.extension('bokeh')

# Load the NETCDF file
file_path = 'tx35bals_CMIP6_ssp126_mon_201501-210012.nc'
data = xr.open_dataset(file_path)

# Select data for the first member and years 2025-2100
member_index = 0
time_2025_2100 = data['time'].sel(time=slice('2025-01-01', '2100-12-31'))
tx35ba_2025_2100 = data['tx35bals'].sel(time=time_2025_2100).isel(member=member_index)

# Define a function to map months to seasons
def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Spring"
    elif month in [6, 7, 8]:
        return "Summer"
    elif month in [9, 10, 11]:
        return "Autumn"

# Assign seasons to the data
tx35ba_2025_2100 = tx35ba_2025_2100.assign_coords(
    season=xr.apply_ufunc(
        get_season,
        tx35ba_2025_2100['time.month'],
        vectorize=True  # Ensures element-wise operation
    )
)

# Group the data by season and year
tx35ba_2025_2100 = tx35ba_2025_2100.assign_coords(year=tx35ba_2025_2100['time.year'])
seasonal_data = tx35ba_2025_2100.groupby(["season", "year"]).sum(dim="time")

# Load country boundaries using GeoPandas
world = gpd.read_file("https://naturalearth.s3.amazonaws.com/110m_cultural/ne_110m_admin_0_countries.zip")

# Convert country boundaries to a GeoViews element
world_gv = gv.Polygons(world, vdims=['NAME']).opts(
    line_color='black', fill_alpha=0.1, line_width=1
)

# Define a function to create a dynamic seasonal plot
def seasonal_plot(season):
    def plot(year):
        data_sel = seasonal_data.sel(season=season, year=year)
        heatmap = data_sel.hvplot(
            x='lon', y='lat', cmap='YlOrRd', clim=(float(seasonal_data.min()), float(seasonal_data.max())),
            geo=True, coastline=True, project=True,
            title=f'{season} {year} Heatmap of tx35ba Variable',
            colorbar=True
        )
        return heatmap * world_gv
    return hv.DynamicMap(plot, kdims=[hv.Dimension('year', values=sorted(seasonal_data.year.values))])

# Create dynamic maps for each season
winter_plot = seasonal_plot("Winter").opts(frame_width=500, frame_height=400)
spring_plot = seasonal_plot("Spring").opts(frame_width=500, frame_height=400)
summer_plot = seasonal_plot("Summer").opts(frame_width=500, frame_height=400)
autumn_plot = seasonal_plot("Autumn").opts(frame_width=500, frame_height=400)

# Combine the seasonal plots into a layout
layout = hv.Layout([winter_plot, spring_plot, summer_plot, autumn_plot]).cols(2)

# Display the interactive layout
layout
```

# We may also just want to look at annual days of heat above 35 degress Celsius

```python
# Import required libraries
import xarray as xr
import hvplot.xarray  # for plotting with hvplot
import geoviews as gv
import geopandas as gpd
import holoviews as hv

# Enable the Bokeh and Geoviews extension for interactive plots
hv.extension('bokeh')

# Load the NETCDF file
file_path = 'tx35bals_CMIP6_ssp126_mon_201501-210012.nc'
data = xr.open_dataset(file_path)

# Select data for the first member and years 2025-2100
member_index = 0
time_2025_2100 = data['time'].sel(time=slice('2025-01-01', '2100-12-31'))
tx35ba_2025_2100 = data['tx35bals'].sel(time=time_2025_2100).isel(member=member_index)

# Group the data by year
tx35ba_2025_2100 = tx35ba_2025_2100.assign_coords(year=tx35ba_2025_2100['time.year'])
annual_data = tx35ba_2025_2100.groupby("year").sum(dim="time")

# Load country boundaries using GeoPandas
world = gpd.read_file("https://naturalearth.s3.amazonaws.com/110m_cultural/ne_110m_admin_0_countries.zip")

# Convert country boundaries to a GeoViews element
world_gv = gv.Polygons(world, vdims=['NAME']).opts(
    line_color='black', fill_alpha=0.1, line_width=1
)

# Define a function to create a dynamic annual plot
def annual_plot(year):
    data_sel = annual_data.sel(year=year)
    heatmap = data_sel.hvplot(
        x='lon', y='lat', cmap='YlOrRd', clim=(float(annual_data.min()), float(annual_data.max())),
        geo=True, coastline=True, project=True,
        title=f'Annual Heatmap of tx35ba Variable ({year})',
        colorbar=True
    )
    return heatmap * world_gv

# Create a DynamicMap for annual data with a slider for years
dynamic_map = hv.DynamicMap(
    annual_plot, 
    kdims=[hv.Dimension('year', values=sorted(annual_data.year.values))]
).opts(frame_width=800, frame_height=600)

# Display the interactive map
dynamic_map
```

## Next, we will add the health facilities to the map

```python
# Import required libraries
import xarray as xr
import hvplot.xarray  # for plotting with hvplot
import geoviews as gv
import geopandas as gpd
import cartopy.crs as ccrs
import holoviews as hv
hv.extension('bokeh', 'matplotlib')

# Load the NETCDF file
file_path = 'tx35bals_CMIP6_ssp126_mon_201501-210012.nc'
data = xr.open_dataset(file_path)

# Select data for the first member and aggregate by year (2025 to 2100)
member_index = 0
time_range = data['time'].sel(time=slice('2025-01-01', '2100-12-31'))
tx35ba_yearly = (
    data['tx35bals']
    .sel(time=time_range)
    .isel(member=member_index)
    .groupby('time.year')
    .sum(dim='time')
)

# Load country boundaries using GeoPandas from a URL
world = gpd.read_file("https://naturalearth.s3.amazonaws.com/110m_cultural/ne_110m_admin_0_countries.zip")

# Convert the country boundaries to a GeoViews element
world_gv = gv.Polygons(world, vdims=['NAME']).opts(
    line_color='black', fill_alpha=0.1, line_width=1
)

# Load the Kuwait-node shapefile for health facilities
gdf_Kuwait = gpd.read_file('Kuwait-node.shp')

# Filter health facilities within Kuwait City's bounding box
xmin, ymin, xmax, ymax = 47, 29, 48, 30
gdf_KuwaitCity = gdf_Kuwait.cx[xmin:xmax, ymin:ymax]

# Import hvplot for geopandas plotting
import hvplot.pandas

# Define a function to plot data for a specific year
def plot_for_year(year):
    # Extract data for the specified year
    tx35ba_year = tx35ba_yearly.sel(year=year)
    
    # Create a heat map for the tx35ba variable
    tx35ba_plot = tx35ba_year.hvplot(
        x='lon', y='lat', clim=(float(tx35ba_yearly.min()), float(tx35ba_yearly.max())),
        cmap='YlOrRd', geo=True, coastline=True, project=True,
        title=f'tx35ba Variable Map - Year {year}',
        tiles='ESRI', colorbar=True,
        xlim=(xmin, xmax), ylim=(ymin, ymax)
    )
    
    # Plot Kuwait City health facilities as points
    kuwait_city_points_plot = gdf_KuwaitCity.hvplot.points(
        x='geometry.x', y='geometry.y', color='red', size=5, hover_cols=['name'],
        geo=True
    )
    
    # Combine the tx35ba plot with country boundaries and health facility points
    overlay = tx35ba_plot * world_gv * kuwait_city_points_plot
    
    # Customize overlay options
    overlay = overlay.opts(
        frame_width=600, frame_height=600,
        title=f'tx35ba Map for 2025-2100 with Kuwait City Health Facilities'
    )
    return overlay

# Create a DynamicMap for interactivity
years = tx35ba_yearly['year'].values.tolist()  # List of years available in the dataset
dynamic_map = hv.DynamicMap(plot_for_year, kdims=['year']).redim.values(year=years)

# Display the dynamic map
dynamic_map
```

# Next, we will perform a spatial join 
-- to join the netcdf4 data on days of deadly heat to the layer of health facilities
and record the number of days of deadly heat in the attribute table of the health facilities layer

```python
import xarray as xr
import geopandas as gpd
import pandas as pd

# Load the NetCDF dataset
file_path = 'tx35bals_CMIP6_ssp126_mon_201501-210012.nc'
ds = xr.open_dataset(file_path)

# Select the year 2025-2100
year = 2025
data_2037 = ds.sel(time=ds.time.dt.year == year)

# Aggregate across the 'member' dimension by taking the mean
aggregated_data_2037 = data_2037['tx35bals'].mean(dim='member', skipna=True)

# Aggregate further by time to get the yearly mean
yearly_mean_2037 = aggregated_data_2037.mean(dim='time', skipna=True)

# Read the point shapefile
points_file = 'Kuwait-node.shp'
points = gpd.read_file(points_file)

# Ensure the shapefile coordinates match the NetCDF CRS (EPSG:4326 is standard for lat/lon)
if points.crs is not None:
    points = points.to_crs(epsg=4326)

# Perform interpolation to assign the yearly mean to points
longitude_array = points.geometry.x.values
latitude_array = points.geometry.y.values

try:
    # Interpolate the values for the points
    interpolated_values = yearly_mean_2037.interp(
        lon=xr.DataArray(longitude_array, dims='points'),
        lat=xr.DataArray(latitude_array, dims='points'),
        method='nearest'
    )
    # Add the interpolated values to the GeoDataFrame
    points[f'tx35bals_{year}'] = interpolated_values.values
except Exception as e:
    print(f"Error during spatial join: {e}")
    points[f'tx35bals_{year}'] = None  # Fallback for failed operations

# Reorder columns to place the new variable next to 'name'
if 'name' in points.columns and f'tx35bals_{year}' in points.columns:
    columns = points.columns.to_list()
    name_index = columns.index('name')
    # Remove the aggregated variable from the current position
    columns.remove(f'tx35bals_{year}')
    # Insert the aggregated variable next to 'name'
    columns.insert(name_index + 1, f'tx35bals_{year}')
    points = points[columns]

# Sort the GeoDataFrame in descending order by the interpolated variable
points = points.sort_values(by=f'tx35bals_{year}', ascending=False)

# Reset index after sorting (optional)
points = points.reset_index(drop=True)

# Display the result
# print(points)
points
```

### Finally, we will aggregate the spatial join results by year to create fields for days above 35c for each year (d35p_2025, d35p_2026, d35p_2027....)

```python
import xarray as xr
import geopandas as gpd
import pandas as pd

# Load the NetCDF dataset
file_path = 'tx35bals_CMIP6_ssp126_mon_201501-210012.nc'
ds = xr.open_dataset(file_path)

# Specify the range of years (e.g., 2025 to 2027)
start_year = 2025
end_year = 2100

# Select the range of years
data_range = ds.sel(time=ds.time.dt.year.isin(range(start_year, end_year + 1)))

# Aggregate across the 'member' dimension by taking the mean
aggregated_data_range = data_range['tx35bals'].mean(dim='member', skipna=True)

# Aggregate further by time to calculate yearly means
yearly_means = aggregated_data_range.groupby('time.year').sum(dim='time', skipna=True)

# Read the point shapefile
points_file = 'Kuwait-node.shp'
points = gpd.read_file(points_file)

# Ensure the shapefile coordinates match the NetCDF CRS (EPSG:4326 is standard for lat/lon)
if points.crs is not None:
    points = points.to_crs(epsg=4326)

# Prepare to add yearly interpolated values to the GeoDataFrame
longitude_array = points.geometry.x.values
latitude_array = points.geometry.y.values

# Loop through each year in the range and interpolate values
for year in range(start_year, end_year + 1):
    try:
        # Extract the data for the specific year
        yearly_data = yearly_means.sel(year=year)
        
        # Interpolate the values for the points
        interpolated_values = yearly_data.interp(
            lon=xr.DataArray(longitude_array, dims='points'),
            lat=xr.DataArray(latitude_array, dims='points'),
            method='nearest'
        )
        
        # Add the interpolated values to the GeoDataFrame
        points[f'd35p_{year}'] = interpolated_values.values
    except Exception as e:
        print(f"Error during interpolation for year {year}: {e}")
        points[f'd35p_{year}'] = None  # Fallback for failed operations

# Optionally, reorder the columns to place yearly variables together
columns_order = ['name'] + [f'd35p_{year}' for year in range(start_year, end_year + 1)] + list(
    set(points.columns) - set([f'd35p_{year}' for year in range(start_year, end_year + 1)] + ['name'])
)
points = points[columns_order]

# Reset index for a clean GeoDataFrame
points = points.reset_index(drop=True)

# Display the resulting GeoDataFrame
# print(points)
points
```
