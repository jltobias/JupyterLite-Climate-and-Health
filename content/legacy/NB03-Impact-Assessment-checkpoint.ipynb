## The Relationship between Climate Change and the HIV/AIDS and TB global syndemic 

- HIV by definition lowers the function of the immune system and people often die of opportunistic infections that take advantage of this immune suppression.
- Tuberculosis is one of the most deadly opportunistic infections.  The HIV and TB pandemics are partners in death and a global syndemic that plagues humanity.
- Research papers are emerging that describe the complex relationships between climate change and the HIV epidemic as well as social determinants of health.

- Greenhouse gases will increase temperatures (global heat stress) and sea-level-rise
- Natural disasters, flooding, and heatwaves will occur with greater frequency and strength over time
- Deforestation, Land degredation, and biodiversity loss will all contribute to depletion of natural resources
- Increased migration, higher infectious disease prevalence, infrastructure erosion, food insecurity are secondary impacts of climate change
- Exacerbation of the HIV/AIDS epidemic are perhaps tertiary impacts as there will be increase risk of HIV disease infections, hospitalizations, and deaths.
- Multiple disease burdens such as co-mordidity with HIV and TB are likely to increase but also other disease burdens such as malaria and cholera may also be a big factor with climate change.
- Household resources are likely to decrease along with decreased availability for the workforce and labor which will exacerbate extreme poverty and lead to worsened health outcomes and more deaths.

# Sea Level Rise is another dimensional component of global heat stress

- The UN Secretary General has described Sea-Level-Rise as a triple threat (reference).   Rising seas will contaminate farm fields, rivers, and waterways that support agriculture and impact food and water insecurity.
- Cholera deaths will also increase due to salt-water intrusion and introduction of marine copepods which carry vibrio cholerae and introduce cholera into rivers, lakes, and other waterways.
- Flooding may also cross contaminate waste-water pits with nearby wells and sources of drinking water which will increase the potential for cholera outbreaks in vulnerable coastal communities.
- Sea-level-rise will increase impacts of cyclones and will result in increased flood related mortality but these rising seas will also damage infrastructure such as healthcare facilities. (Tobias et. al, 2024).
- The HIV and TB epidemics will not be immune to sea-level-rise impacts as many PEPFAR-Supported healthcare facilities may be impacted in vulnerable coastal cities in many of the PEFPAR-supported countries (Tobias et. al, 2024).

### Let's use the cdsapi to download a zipped file with several netcdf4 files with the sea-level-rise data from 2025 - 2100

```python
# This is example code to retrieve sea-level-rise data from Copernicus Climate Data Store API
import cdsapi

dataset = "sis-water-level-change-indicators-cmip6"
request = {
    "derived_variable": [
        "absolute_change",
        "absolute_value",
        "percentage_change"
    ],
    "product_type": ["single_model"],
    "model": ["cmcc_cm2_vhr4"],
    "confidence_interval": [
        "best_fit",
        "low_bound_confidence_interval",
        "high_bound_confidence_interval"
    ],
    "experiment": ["future"],
    "period": ["2025"],
    "variable": [
        "annual_mean_of_highest_high_water",
        "annual_mean_of_lowest_low_water",
        "highest_astronomical_tide",
        "lowest_astronomical_tide",
        "mean_sea_level",
        "tidal_range"
    ]
}

# Define the output ZIP file name
zip_file_name = "future_tide_actual-value-cmcc_cm2_vhr4_2025-2100.zip"

# initialize client
client = cdsapi.Client()

# Download the file using cdsapi
client.retrieve(dataset, request, zip_file_name)


# client.retrieve(dataset, request).download()
```

#### Let's extract the netcdf4 file from the zip file

```python
# extract the netcdf4 from the zipfile

# import libraries 
import os
import zipfile

# Check if the zipfile has been downloaded
if not os.path.exists(zip_file_name):
    print(f"Error: File '{zip_file_name}' not found.")
    exit()

# 2. Check if the file is a ZIP archive and extract it if necessary
if zipfile.is_zipfile(zip_file_name):
    with zipfile.ZipFile(zip_file_name, 'r') as zip_ref:
        zip_ref.extractall(".")  # Extract contents to the current directory
    # Find the extracted NetCDF file
    extracted_files = [f for f in os.listdir(".") if f.endswith('.nc')]
    if not extracted_files:
        print("Error: No NetCDF (.nc) file found in the extracted archive.")
        exit()
    else:
        file_name = extracted_files[0]  # Use the first extracted NetCDF file
        print(f"NetCDF file extracted: {file_name}")
else:
    print(f"Error: The file '{zip_file_name}' is not a valid ZIP archive.")
    exit()
```

### We may want to do this for any / all zipfiles in the directory 

```python
import os
import zipfile

# Function to process all ZIP files in the current directory
def process_all_zip_files():
    # List all ZIP files in the current directory
    zip_files = [f for f in os.listdir(".") if f.endswith('.zip')]
    
    if not zip_files:
        print("Error: No ZIP files found in the current directory.")
        return

    extracted_nc_files = []  # List to store the names of all extracted NetCDF files
    
    for zip_file_name in zip_files:
        print(f"Processing ZIP file: {zip_file_name}")
        # Check if the file is a ZIP archive
        if zipfile.is_zipfile(zip_file_name):
            with zipfile.ZipFile(zip_file_name, 'r') as zip_ref:
                # Extract all contents to the current directory
                zip_ref.extractall(".")
            
            # Find the extracted NetCDF files
            extracted_files = [f for f in os.listdir(".") if f.endswith('.nc')]
            if not extracted_files:
                print(f"Warning: No NetCDF (.nc) files found in the archive '{zip_file_name}'.")
            else:
                # Append the names of extracted NetCDF files to the list
                extracted_nc_files.extend(extracted_files)
                print(f"NetCDF files extracted from '{zip_file_name}': {extracted_files}")
        else:
            print(f"Error: The file '{zip_file_name}' is not a valid ZIP archive.")
    
    # Print all extracted NetCDF files
    if extracted_nc_files:
        print("\nAll extracted NetCDF files:")
        for nc_file in extracted_nc_files:
            print(nc_file)
    else:
        print("No NetCDF (.nc) files were extracted from the ZIP archives.")

# Run the function
process_all_zip_files()
```

### Next - let's try to visualize the sea-level-rise data on a global map

```python
import hvplot.xarray  # For plotting with hvplot
import geoviews as gv
import geopandas as gpd
import holoviews as hv
import xarray as xr
import cartopy.crs as ccrs

# Enable extensions for interactive plots
hv.extension('bokeh')

# Load the NETCDF file
file_path = 'future_tide_actual-value_2025_MSL_v1.nc'
data = xr.open_dataset(file_path)

# Load country boundaries using GeoPandas
world = gpd.read_file("https://naturalearth.s3.amazonaws.com/110m_cultural/ne_110m_admin_0_countries.zip")

# Convert the country boundaries to a GeoViews element
world_gv = gv.Polygons(world, vdims=['NAME']).opts(
    line_color='black', fill_alpha=0.1, line_width=1, projection=ccrs.PlateCarree()
)

# Extract the Mean Sea Level (MSL) variable and station coordinates
msl = data['MSL'].isel(time=0).data  # Single time index
lon = data['station_x_coordinate'].data  # Extract longitude
lat = data['station_y_coordinate'].data  # Extract latitude

# Convert the data into a GeoViews Points element
points = gv.Points((lon, lat, msl), vdims='MSL', kdims=['Longitude', 'Latitude']).opts(
    color='MSL', cmap='YlOrRd', colorbar=True, size=5, tools=['hover'],
    clim=(msl.min(), msl.max()), projection=ccrs.PlateCarree()
)

# Overlay the country boundaries on the MSL plot
overlay = points * world_gv

# Configure the overlay display
overlay.opts(
    frame_width=1200, frame_height=800,
    title='Mean Sea Level (2025) with Country Boundaries'
)

# Display the overlay
overlay
```
