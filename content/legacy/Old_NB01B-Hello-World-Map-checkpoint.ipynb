```python
# Step 1: Import required libraries
import xarray as xr
import hvplot.xarray  # for plotting with hvplot
import geoviews as gv
import geopandas as gpd
import cartopy.crs as ccrs

# Enable the Bokeh and Geoviews extension for interactive plots
import holoviews as hv
hv.extension('bokeh', 'matplotlib')

# Load the NETCDF file
file_path = 'tx35ba_CMIP6_ssp126_mon_201501-210012.nc'
data = xr.open_dataset(file_path)

# Step 2: Select data for the first member and year 2015
member_index = 0
time_2015 = data['time'].sel(time=slice('2015-01-01', '2015-12-31'))
tx35ba_2015 = data['tx35ba'].sel(time=time_2015).isel(member=member_index)

# Load country boundaries using GeoPandas from a URL
world = gpd.read_file("https://naturalearth.s3.amazonaws.com/110m_cultural/ne_110m_admin_0_countries.zip")

# Convert the country boundaries to a GeoViews element
world_gv = gv.Polygons(world, vdims=['NAME']).opts(
    line_color='black', fill_alpha=0.1, line_width=1  # Replace edge_line_width with line_width
)

# Step 3: Create interactive time-series map using hvplot with boundaries

# Plotting the data with hvplot
tx35ba_plot = tx35ba_2015.hvplot(
    x='lon', y='lat', clim=(float(tx35ba_2015.min()), float(tx35ba_2015.max())),
    cmap='YlOrRd', geo=True, coastline=True, project=True,
    title='tx35ba Variable Map - Time Series for 2015',
    tiles='ESRI', colorbar=True  # Move colorbar=True here
)

# Overlay the country boundaries
overlay = tx35ba_plot * world_gv

# Display the overlay
overlay.opts(
    frame_width=800, frame_height=600,
    title='tx35ba Time-Series Map for 2015 with Country Boundaries'
)


```

## Next, let us create the heat map for the CMIP6 data for the RCP 2.6 scenario:

```python
import xarray as xr
import folium
from folium.plugins import HeatMap
import numpy as np
import zipfile
import os

# Path to the .nc file
nc_file_path = 'tx35ba_CMIP6_ssp126_mon_201501-210012.nc'

# Load the .nc file
nc_data = xr.open_dataset(nc_file_path)

# Select the first time point and first model member
data_slice = nc_data['tx35ba'].isel(time=0, member=0)

# Create a base map centered on a rough global midpoint
m = folium.Map(location=[0, 0], zoom_start=3)

# Prepare data for HeatMap (must be in the form of (lat, lon, value))
heat_data = [[lat, lon, data_slice.sel(lat=lat, lon=lon).values.item()] 
             for lat in data_slice.lat.values 
             for lon in data_slice.lon.values 
             if not np.isnan(data_slice.sel(lat=lat, lon=lon).values.item())]

# Create color gradient
#gradient = {.33: 'red', .66: 'orange', 1: 'yellow'}

gradient = {.33: 'red', .66: 'orange', 1: 'yellow'}



# Create and add the HeatMap layer
heatmap = HeatMap(heat_data, gradient=gradient, radius=5)
m.add_child(heatmap)




# Save the map as an HTML file
map_path = 'heatmap.html'
m.save(map_path)
print(f"Heatmap saved at {map_path}")

m
```

```python
import xarray as xr
import folium
from folium.plugins import HeatMapWithTime
import numpy as np

# Path to the .nc file
nc_file_path = 'tx35ba_CMIP6_ssp126_mon_201501-210012.nc'

# Load the .nc file
nc_data = xr.open_dataset(nc_file_path)

# Create a base map centered on a rough global midpoint
m = folium.Map(location=[0, 0], zoom_start=3)

# List to hold heatmap data for each year
heat_data_by_year = []

# Extract unique years from the time coordinate
time_index = nc_data['time'].dt.year.values
unique_years = np.unique(time_index)

# Loop through each year, extract the relevant data slice, and prepare the heatmap data
for year in unique_years:
    # Select data for the year (assuming monthly data, we can take the mean of the year)
    yearly_data = nc_data['tx35ba'].sel(time=str(year)).mean(dim='time')
    
    # Prepare heatmap data for this year (lat, lon, value)
    heat_data = []
    for lat in yearly_data.lat.values:
        for lon in yearly_data.lon.values:
            # Extract scalar value for each lat-lon point and ensure it's a single element array
            value = yearly_data.sel(lat=lat, lon=lon).values
            
            # Check if it's a scalar and handle NaN values
            if np.isscalar(value) and not np.isnan(value):
                heat_data.append([lat, lon, float(value)])  # Ensure it's a float scalar

    heat_data_by_year.append(heat_data)

# Add HeatMapWithTime to the map
HeatMapWithTime(
    data=heat_data_by_year,
    index=unique_years.tolist(),  # Add years as the time slider index
    radius=5,
    auto_play=False,
    max_opacity=0.6
).add_to(m)

# Save the map as an HTML file
map_path = 'heatmap_with_slider.html'
m.save(map_path)
print(f"Heatmap with time slider saved at {map_path}")

m
```
